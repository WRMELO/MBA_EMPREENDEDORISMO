```python
import os
import pandas as pd
from datetime import datetime

# Caminhos oficiais
base_dir = r"C:\Users\wilso\MBA_EMPREENDEDORISMO\3AGD"
saida_csv = os.path.join(base_dir, "desenvolvimento", "inventario_arquivos.csv")

# Lista para armazenar resultados
dados = []

# Percorre todas as pastas e arquivos recursivamente
for raiz, dirs, files in os.walk(base_dir):
    for nome in dirs:
        caminho = os.path.join(raiz, nome)
        stats = os.stat(caminho)
        dados.append({
            "nome_arquivo": nome,
            "tipo": "pasta",
            "caminho_completo": caminho,
            "ultima_modificacao": datetime.fromtimestamp(stats.st_mtime),
            "tamanho_bytes": stats.st_size
        })
    for nome in files:
        caminho = os.path.join(raiz, nome)
        stats = os.stat(caminho)
        dados.append({
            "nome_arquivo": nome,
            "tipo": "arquivo",
            "caminho_completo": caminho,
            "ultima_modificacao": datetime.fromtimestamp(stats.st_mtime),
            "tamanho_bytes": stats.st_size
        })

# Cria diretório de saída se não existir
os.makedirs(os.path.dirname(saida_csv), exist_ok=True)

# Salva o inventário em CSV
df = pd.DataFrame(dados)
df.to_csv(saida_csv, index=False, encoding="utf-8-sig")

print(f"Inventário gerado com sucesso: {saida_csv}")
```

```python
import os
import shutil
import pandas as pd

base = r"C:\Users\wilso\MBA_EMPREENDEDORISMO\3AGD"

# Estrutura alvo organizada
estrutura = [
    "desenvolvimento",
    "a3_backend_local\\src",
    "a3_backend_local\\config",
    "a3_backend_local\\docker",
    "a3_backend_local\\data",
    "a3_backend_docker",
    "a3_backend_docker\\volumes",
    "docs",
    "legacy"
]

# Criar estrutura
for d in estrutura:
    os.makedirs(os.path.join(base, d), exist_ok=True)

def gerar_nome_unico_dir(destino):
    """Se a pasta já existir, cria uma versão _vN."""
    if not os.path.exists(destino):
        return destino
    v = 1
    while os.path.exists(f"{destino}_v{v}"):
        v += 1
    return f"{destino}_v{v}"

# Função para mover diretórios com versionamento
def mover_diretorio(origem, destino_base):
    destino = os.path.join(destino_base, os.path.basename(origem))
    destino_final = gerar_nome_unico_dir(destino)
    shutil.move(origem, destino_final)
    return destino_final

# Listar todos os diretórios de primeiro nível que não fazem parte da estrutura padrão
dirs_top = [os.path.join(base, d) for d in os.listdir(base)
            if os.path.isdir(os.path.join(base, d)) and d not in [p.split("\\")[0] for p in estrutura]]

movidos = []
for d in dirs_top:
    try:
        destino = mover_diretorio(d, os.path.join(base, "legacy"))
        movidos.append((d, destino))
    except Exception as e:
        print(f"Erro movendo {d}: {e}")

# Criar log de movimentação
df = pd.DataFrame(movidos, columns=["origem", "destino"])
log_csv = os.path.join(base, "desenvolvimento", "log_reordenacao_diretorios.csv")
df.to_csv(log_csv, index=False, encoding="utf-8-sig")

print(f"Reorganização concluída no nível de diretórios. Log salvo em {log_csv}")
```

```python
import os
import shutil
import pandas as pd

# Caminho base
base = r"C:\Users\wilso\MBA_EMPREENDEDORISMO\3AGD"
venv_principal = os.path.join(base, ".venv")

deletados = []

# Percorrer recursivamente
for raiz, dirs, _ in os.walk(base, topdown=False):
    for d in dirs:
        if d == ".venv":
            caminho = os.path.join(raiz, d)
            if os.path.abspath(caminho) != os.path.abspath(venv_principal):
                try:
                    shutil.rmtree(caminho)
                    deletados.append(caminho)
                    print(f"Removido: {caminho}")
                except Exception as e:
                    print(f"Erro ao remover {caminho}: {e}")

# Gerar log
log_csv = os.path.join(base, "desenvolvimento", "log_venv_deletados.csv")
os.makedirs(os.path.dirname(log_csv), exist_ok=True)
pd.DataFrame(deletados, columns=["venv_removido"]).to_csv(log_csv, index=False, encoding="utf-8-sig")

print(f"Limpeza concluída. Log salvo em: {log_csv}")
```

```python
import os
import pandas as pd
from datetime import datetime

# Diretório base contendo os arquivos .md
base_dir = r"C:\Users\wilso\MBA_EMPREENDEDORISMO\3AGD\A3_LOCAL\data"

# Lista para inventário
dados = []

# Percorrer recursivamente
for raiz, _, arquivos in os.walk(base_dir):
    for f in arquivos:
        if f.lower().endswith(".md"):
            caminho = os.path.join(raiz, f)
            stats = os.stat(caminho)
            dados.append({
                "arquivo": f,
                "caminho_completo": caminho,
                "tamanho_bytes": stats.st_size,
                "ultima_modificacao": datetime.fromtimestamp(stats.st_mtime)
            })

# Criar DataFrame
df = pd.DataFrame(dados)

# Salvar inventário
saida_csv = os.path.join(base_dir, "..", "desenvolvimento", "inventario_md.csv")
os.makedirs(os.path.dirname(saida_csv), exist_ok=True)
df.to_csv(saida_csv, index=False, encoding="utf-8-sig")

print(f"Inventário gerado com sucesso: {saida_csv}")
```
